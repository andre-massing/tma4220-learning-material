```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 1, 400)
u = np.sin(2 * np.pi * x) * np.exp(-x)

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(x, u, label=r"$u(x) = \sin(2\pi x)\,e^{-x}$")
ax.axhline(0, color="black", linewidth=0.8)
ax.set_xlabel("$x$")
ax.set_ylabel("$u(x)$")
ax.legend()
ax.grid(alpha=0.3)
fig.tight_layout()
```
